## Parâmetros unitários

```python
z1 = 0.01539 + 1j * 0.25163 # Series impedance per unit length (ohms/km)
y1 = 0.003e-6 + 1j * 6.5265e-6 # Shunt admittance per unit length (S/km)
ell = 300.0 # comprimento do circuito (km)
gamma = np.sqrt(z1 * y1) # Propagation constant (1/m)
theta = gamma * ell # Electrical length of the line (rad)
theta_i = np.sqrt(np.imag(z1) * np.imag(y1)) * ell # Electrical length of the line (rad)
print(f"theta_i = {theta_i:.4f} rad")
```

```python
# verificacao de sanidade
theta_i/np.pi * 2500
```

```python
# compensacao de reativos
cs = 0.3 # compensacao em serie 
cd = 0.75 # compensacao em derivação
eta = 1 - cs # fator de compensação em derivação
xi = 1 - cd # fator de compensação em série
# calculo do novo comprimento eletrico da linha
theta_c = np.sqrt(eta * xi) * theta_i
print(f"theta_c = {theta_c:.4f} rad")
```

```python
# em km 
theta_c/np.pi * 2500
```
